```python
# Create sample data

data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Headphone", "Watch"],
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [500, 700, 450, 800, 600],
    "Profit": [100, 150, 80, 200, 120],
    "Marketing": [50, 70, 40, 90, 60]
}
```

```python
# Create DataFrame

dataset = pd.DataFrame(data)

# Display the dataset
print(dataset)
```

```python
plt.bar(dataset["Product"], dataset["Sales"])
plt.show()
```

```python
# Bar Chart
# Shows sales for each product

plt.bar(dataset["Product"], dataset["Sales"])

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product-wise Sales")

plt.show()
```

```python
# Line Chart
# Shows sales trend across months

plt.plot(
    dataset["Month"],
    dataset["Sales"],
    marker='o'
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
```

```python
# Pie Chart
# Shows the percentage contribution of each product

plt.pie(
    dataset["Sales"],
    labels=dataset["Product"],
    autopct="%1.1f%%"
)

plt.title("Sales Distribution by Product")

plt.show()
```

```python
plt.hist(dataset["Sales"], bins=3)
```

```python
# Histogram
# Shows the distribution of sales values

plt.hist(
    dataset["Sales"],
    bins=3
)

plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.title("Sales Distribution")

plt.show()
```

```python
# Scatter Plot
# Shows relationship between Marketing and Sales

plt.scatter(
    dataset["Marketing"],
    dataset["Sales"]
)

plt.xlabel("Marketing")
plt.ylabel("Sales")
plt.title("Marketing vs Sales")

plt.show()
```

```python
# Box Plot
# Shows the distribution and possible outliers in Sales

plt.boxplot(dataset["Sales"])

plt.ylabel("Sales")
plt.title("Sales Distribution")

plt.show()
```

```python
# Horizontal Bar Chart

plt.barh(
    dataset["Product"],
    dataset["Sales"]
)

plt.xlabel("Sales")
plt.ylabel("Product")
plt.title("Product-wise Sales")

plt.show()
```

```python
# Area Chart

x = range(len(dataset["Month"]))

plt.fill_between(
    x,
    dataset["Sales"]
)

plt.xticks(
    x,
    dataset["Month"]
)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Area Chart")

plt.show()
```

```python
# Stack Plot
# Shows Sales and Profit together

x = range(len(dataset["Month"]))

plt.stackplot(
    x,
    dataset["Sales"],
    dataset["Profit"],
    labels=["Sales", "Profit"]
)

plt.xticks(
    x,
    dataset["Month"]
)

plt.xlabel("Month")
plt.ylabel("Value")
plt.title("Sales and Profit")

plt.legend()

plt.show()
```

```python
# Create data for heatmap

heatmap = [
    dataset["Sales"],
    dataset["Profit"],
    dataset["Marketing"]
]
```

```python
# Heatmap

plt.imshow(heatmap)

plt.xticks(
    range(5),
    dataset["Month"]
)

plt.yticks(
    range(3),
    ["Sales", "Profit", "Marketing"]
)

plt.colorbar()

plt.title("Sales, Profit and Marketing Heatmap")

plt.show()
```
